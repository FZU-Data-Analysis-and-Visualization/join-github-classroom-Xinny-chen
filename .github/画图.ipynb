```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0, 10, 100)  
y=np.sin(x)  
y2=np.cos(x)            
plt.plot(x, y,'r--',marker='o',label='$\\sin(x)$')
plt.plot(x, y2)
plt.title("$\\sin(x)$ and $\\cos(x)$") 
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

x轴范围：plot.xlim（）
水平线：plot.axhline()
显示点plt.plot(x, y, marker='o')
网格plt.grid()
保存图片plt.savefig("sin_plot.png")

```python
import matplotlib.pyplot as plt

labels = ["Research", "Design", "Development", "Testing", "Deployment"]
sizes = [28, 18, 32, 14, 8]

colors = [
    "#264653",  # 深青蓝
    "#2A9D8F",  # 高级绿
    "#E9C46A",  # 柔和黄
    "#F4A261",  # 暖橙
    "#E76F51"   # 珊瑚红
]

plt.figure(figsize=(8, 8), dpi=150)

wedges, texts, autotexts = plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    counterclock=False,
    pctdistance=0.72,
    labeldistance=1.08,
    wedgeprops={
        "edgecolor": "white",
        "linewidth": 2
    },
    textprops={
        "fontsize": 12,
        "color": "#222222"
    }
)

for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontsize(11)
    autotext.set_fontweight("bold")

plt.title(
    "Project Resource Allocation",
    fontsize=18,
    fontweight="bold",
    color="#1F2933",
    pad=20
)

plt.axis("equal")
plt.tight_layout()
plt.show()
```
