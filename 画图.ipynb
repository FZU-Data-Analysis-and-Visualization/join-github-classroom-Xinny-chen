```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0, 10, 100)  
y=np.sin(x)              
plt.plot(x, y)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("y=sin(x)")
plt.show()
```

x轴范围：plot.xlim（）
水平线：plot.axhline()
显示点plt.plot(x, y, marker='o')
网格plt.grid()
保存图片plt.savefig("sin_plot.png")
