
### 六、合并与拼接

1. 横向合并：

   df1 = pd.DataFrame ({"id":[1,2], "name":["X","Y"]})

   df2 = pd.DataFrame ({"id":[1,2], "age":[20,30]})

   按 id 合并。

   

2. 纵向拼接：

   df_a = pd.DataFrame ({"val":[1,2]})

   df_b = pd.DataFrame ({"val":[3,4]})

   拼接后重置索引。

```python
import pandas as pd
df1 = pd.DataFrame ({"id":[1,2], "name":["X","Y"]})

df2 = pd.DataFrame ({"id":[1,2], "age":[20,30]})
df1,df2
```

```python
result=pd.concat([df1,df2],join='outer',axis=1)
result
```

```python
import pandas as pd
df_a = pd.DataFrame ({"val":[1,2]})
df_b = pd.DataFrame ({"val":[3,4]})
result=pd.concat([df_a,df_b],axis=0,ignore_index=True)
result
```

### 八、高级操作

1. 对 score 列做标准化（(值 - 均值)/ 标准差）。

   

2. 用透视表：行是 gender，值是 score，聚合函数为中位数。

```python
import pandas as pd
df = pd.DataFrame({
    'name': ['Tom', 'Jerry', 'Mike'], 'score': [85, 92, 78],'gender': ['男', '女', '男']
})
df
```

```python
df['score_std'] = (df['score'] - df['score'].mean()) / df['score'].std()
df
```

```python
pivot_result = df.pivot_table(
    index='gender',
    values='score',
    aggfunc='median'
)
pivot_result
```
